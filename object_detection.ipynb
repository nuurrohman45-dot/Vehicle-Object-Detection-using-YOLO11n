```python
from ultralytics import YOLO
import cv2 as cv
import matplotlib
%matplotlib inline
import matplotlib.pyplot as plt
```

```python
import torch
print(torch.cuda.is_available())
```

# Pretrained Model

```python
pre_model = YOLO("yolo11n.pt") 
```

```python
# Prediction
img_path = r"datasets\images\val\17d0db0b-val_img_5.jpg"

results = pre_model.predict(source=img_path, conf=0.5, verbose=False)

img_bgr = results[0].plot()
img_rgb = cv.cvtColor(img_bgr, cv.COLOR_BGR2RGB)

plt.figure(figsize=(10, 6))
plt.imshow(img_rgb)
plt.title("YOLO Prediction (Ultralytics plot)")
plt.axis("off")
plt.show()
```

```python
# Prediction
img_path = r"datasets\images\val\7d5dafe4-val_img_15.jpg"

results = pre_model.predict(source=img_path, conf=0.5, verbose=False)

img_bgr = results[0].plot()
img_rgb = cv.cvtColor(img_bgr, cv.COLOR_BGR2RGB)

plt.figure(figsize=(10, 6))
plt.imshow(img_rgb)
plt.title("YOLO Prediction (Ultralytics plot)")
plt.axis("off")
plt.show()
```

# Trained Model

```python
model = YOLO("yolo11n.pt") 

results = model.train(
    data=r"datasets\data.yaml",   
    epochs=50,
    imgsz=640,
    batch=4,                      
    device="cpu"                   
)
```

```python
# Best model
best_model = YOLO("runs/detect/train2/weights/best.pt")
```

```python
# Validation
best_model.val(data=r"datasets/data.yaml")
```

```python
# Prediction
img_path = r"datasets\images\val\17d0db0b-val_img_5.jpg"

results = best_model.predict(source=img_path, conf=0.5, verbose=False)

img_bgr = results[0].plot()
img_rgb = cv.cvtColor(img_bgr, cv.COLOR_BGR2RGB)

plt.figure(figsize=(10, 6))
plt.imshow(img_rgb)
plt.title("YOLO Prediction (Ultralytics plot)")
plt.axis("off")
plt.show()
```
